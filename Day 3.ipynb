ndim
dimensions ki count batata hai. Ye total elements ki count nahi hai.

```python
#0-D array-Scaler 

a = np.array(10)
print(a.ndim)  # 0

#1-D array-Vector

a = np.array([10, 20, 30])
print(a.ndim)  # 1

#2-D array-Two -dimensional

a = np.array([[1, 2], [3, 4]])
print(a.ndim)  # 2
```

Dtype


dtype ka matlab hai data type.

```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([1.5, 2.5, 3.5])
c = np.array([True, False, True])

print(a.dtype)
print(b.dtype)
print(c.dtype)


#dtype=np.float64 
# specify karne se values floating-point format mein store hongi.

a = np.array([10, 20, 30], dtype=np.float64)

print(a)
print(a.dtype)
```

Aggregation Operations 

```python
# Creating a NumPy array
a = np.array([10, 20, 30, 40, 50])

# 1. SUM: Adds all elements of the array
print("Sum:", np.sum(a))
# Output: 150

# 2. MEAN: Calculates the average of all elements
print("Mean:", np.mean(a))
# Output: 30.0

# 3. MEDIAN: Finds the middle value
print("Median:", np.median(a))
# Output: 30.0

# 4. MIN: Finds the smallest element
print("Minimum:", np.min(a))
# Output: 10

# 5. MAX: Finds the largest element
print("Maximum:", np.max(a))
# Output: 50

# 6. STANDARD DEVIATION: Measures how spread out
# the values are around the mean
print("Standard Deviation:", np.std(a))
# Output: 14.142135623730951

# 7. VARIANCE: Measures the average squared
# distance from the mean
print("Variance:", np.var(a))
# Output: 200.0

# 8. ARGMIN: Returns the index of the smallest element
print("Index of Minimum:", np.argmin(a))
# Output: 0

# 9. ARGMAX: Returns the index of the largest element
print("Index of Maximum:", np.argmax(a))
# Output: 4

# 10. PRODUCT: Multiplies all elements together
print("Product:", np.prod(a))
# Output: 12000000

# 11. CUMSUM: Calculates the cumulative sum
print("Cumulative Sum:", np.cumsum(a))
# Output: [ 10  30  60 100 150]

# 12. CUMPROD: Calculates the cumulative product
print("Cumulative Product:", np.cumprod(a))
# Output: [     10     200    6000  240000 12000000]

# 13. PERCENTILE: Finds the value at a given percentile
print("50th Percentile:", np.percentile(a, 60))
# Output: 30.0

# 14. QUANTILE: Finds the value at a given quantile
print("Quantile:", np.quantile(a, 0.5))
# Output: 30.0

# 15. PTp: Finds the range (maximum - minimum)
print("Range:", np.ptp(a))
# Output: 40
```

Numerical Operations

```python
# Creating two NumPy arrays
a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

# 1. ADDITION: Adds corresponding elements
print("Addition:", a + b)
# Output: [12 24 35]

# 2. SUBTRACTION: Subtracts corresponding elements
print("Subtraction:", a - b)
# Output: [ 8 16 25]

# 3. MULTIPLICATION: Multiplies corresponding elements
print("Multiplication:", a * b)
# Output: [ 20  80 150]

# 4. DIVISION: Divides corresponding elements
print("Division:", a / b)
# Output: [5. 5. 6.]

# 5. FLOOR DIVISION: Divides and rounds down
print("Floor Division:", a // b)
# Output: [5 5 6]

# 6. MODULUS: Returns the remainder after division
print("Modulus:", a % b)
# Output: [0 0 0]

# 7. POWER: Raises each element of a to the
# corresponding power in b
print("Power:", a ** b)
# Output: [     100  160000 24300000]

# 8. SQUARE ROOT: Calculates the square root
print("Square Root:", np.sqrt(a))
# Output: [3.16227766 4.47213595 5.47722558]

# 9. SQUARE: Squares every element
print("Square:", np.square(a))
# Output: [100 400 900]

# 10. ABSOLUTE VALUE: Converts negative values
# into non-negative values
c = np.array([-10, -20, 30])
print("Absolute Value:", np.abs(c))
# Output: [10 20 30]

# 11. EXPONENTIAL: Calculates e raised to each element
print("Exponential:", np.exp(np.array([0, 1, 2])))
# Output: [1.         2.71828183 7.3890561 ]

# 12. LOGARITHM: Calculates the natural logarithm
# of each positive element
print("Natural Log:", np.log(a))
# Output: [2.30258509 2.99573227 3.40119738]

# 13. ROUNDING: Rounds decimal values
d = np.array([2.35, 4.67, 6.12])
print("Rounded:", np.round(d, 1))
# Output: [2.4 4.7 6.1]

# 14. COMPARISON: Compares corresponding elements
print("Greater Than:", a > b)
# Output: [ True  True  True]

# 15. TRIGONOMETRY: Calculates sine values
# Input values are in radians
print("Sine:", np.sin(np.array([0, np.pi / 2, np.pi])))
# Output: [0.0000000e+00 1.0000000e+00 1.2246468e-16]

# 16. CLIP: Restricts values to a minimum and maximum
print("Clipped:", np.clip(a, 15, 25))
# Output: [15 20 25]
```
