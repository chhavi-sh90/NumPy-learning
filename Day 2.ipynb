Array Attributes

```python
import numpy as np
```

```python
a1=np.arange(10)
a2=np.arange(12,dtype=float).reshape(3,4)
a3=np.arange(8).reshape(2,2,2)
a3
```

#number of dimensions(ndim)

```python
print(a3.ndim)
print(a2.ndim)
```

#shape
(it tells how many rows or columns matrix have !)

```python
print(a1.shape)
print(a2.shape)
print(a3.shape)
```

#size(tells how many items matrix have)

```python
print(a1.size)
print(a2.size)
print(a3.size)
```

#itemsize(tells each item occupies how much space in memory)

```python
print(a1.itemsize)
print(a2.itemsize)
print(a3.itemsize)
```

dtype(data type)

```python
print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
```

Changing Datatype

```python
#dtype conversion 
a3.astype(np.int32)
a1.astype(np.float64)
```

Array Operations

```python
a1=np.arange(12).reshape(4,3)
a2=np.arange(12,24).reshape(4,3)
print(a1)
print(a2)
```

```python
#scalar operations
#arithmetic operations

print(a2*0)
print(a1-a2)
print(a1+a2)
```

```python
#relationaal operations
print(a1<a2)
print(a1==a2)
print(a1!=a2)
```

```python
#vector operations
print(a1/a2)
```

Array Functions

```python
a1=np.random.random((3,3))
a1=np.round(a1*100)
a1
```

```python
#max/min/sum/prod
#0=column,1=row
print(np.max(a1))
print(np.min(a1))
print(np.sum(a1))
print(np.prod(a1))
print(np.max(a1, axis=0))
print(np.min(a1, axis=1))
print(np.sum(a1, axis=0))
print(np.prod(a1, axis=1))
```

```python
#mean/median/std/var
print(np.mean(a1))
print(np.median(a1))
print(np.std(a1, axis=0))
print(np.var(a1, axis=1))   
```

```python
#trignometric functions
np.tan(a1)
```

```python
# dot product
a2=np.arange(12).reshape(3,4)
a3=np.arange(12,24).reshape(4,3)
np.dot(a2,a3)
```

```python
#log and exponential functions
np.log(a1)
np.exp(a1) 
```

```python
#round/floor/ceil
a4=np.random.random((3,3))  
print(np.round(a4*100))
print(np.floor(a4))
print(np.ceil(a4))  
```

Indexing and Slicing

```python
a1=np.arange(10)
a2=np.arange(12).reshape(3,4)
a3=np.arange(8).reshape(2,2,2)
print(a2)


```

```python
#indexing
print(a2)
a2[1,2]
```

```python
print(a3)
a3[1,1,0]
```

```python
#slicing
print(a1)
a1[2:5]
```

```python
print(a2)
print(a2[1:3, 0:2])
print(a2[0,:])
print(a2[:,3])
print(a2[1:3,1:3])
```

```python
print(a2[::2,::3])
print(a2[::2,1::2])
print(a2[1,::3])
print(a2[0:2,1:4])
```

```python
a3=np.arange(27).reshape(3,3,3)
print(a3)
print(a3[1,::])
print(a3[0,1,0:4])
print(a3[1,::,1])
print(a3[2,1:3,1:3])
```

Iterating


```python
print(a1)
for i in a1:
    print(i)    
```

```python
print(a2)
for i in a2:
    print(i)    
```

```python
print(a3)
for i in a3:
    print(i)        
```

```python
for i in np.nditer(a3):
    print(i)    
```

Stacking

```python
#horizontal stacking
a4=np.arange(12).reshape(3,4)
a5=np.arange(12,24).reshape(3,4)    
np.hstack((a4,a5,a5))
```

```python
#vertical stacking  
np.vstack((a4,a5,a5))
```

Splitting

```python
#horizontal stacking
np.hsplit(a4,2)
```

```python
#vertical stacking
np.vsplit(a4,3)
```
